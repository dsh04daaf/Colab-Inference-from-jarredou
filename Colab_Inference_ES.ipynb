## ¿Qué modelo usa cada opción?
Los recomendados son **los mejores modelos públicos de cada tipo según la [guía de deton24](https://docs.google.com/document/d/17fjNvJzj8ZGSer7c7OFe_CNfUKbAxEh_OBv94ZdRG5c)** (edición del 10-11-2026), que se basa en métricas y en lo que dicen sus autores y la comunidad:

| Opción | Modelo | Por qué |
|---|---|---|
| Acapella | **Leap Xe** (unwa) | las mejores métricas de voces entre los modelos públicos |
| Instrumental | **deux** (becruily) | el mejor equilibrio entre cuerpo y limpieza; no necesita phase fix |
| Stems | **BS-Roformer SW** (6 stems) | el mejor SDR de un solo modelo para batería, bajo, guitarra y piano |
| Batería por piezas | SW → **DrumSep 5 stems** (jarredou) | el DrumSep público más nuevo y mejor |
| Karaoke | **BS-Roformer** (becruily & frazer) | el que mejor separa voz principal de coros |
| Quitar reverb | **BS-Roformer dereverb** (anvuew) | «probablemente el mejor por ahora» |
| Quitar ruido | **denoise/debleed** (Gabox) | el primero de los de uso general; salen `_denoised` y `_noise` |

No hay un mejor modelo para todo: depende de la canción. Si uno no te convence, prueba otro en *Avanzado → modelo*. Por ejemplo, para instrumentales con más cuerpo, **HyperACE v2** (unwa); para voces, **big_beta7** o **HyperACE v2**.

## ¿En qué orden está la lista de modelos?
Dentro de cada categoría (VOCALS, INST, KARAOKE, DE-REVERB…) van **de mejor a peor en teoría**, en el orden en que los presenta la guía de deton24. Los que la guía no ordena van después, del más nuevo al más viejo. «Peor» solo quiere decir que suele puntuar menos: en una canción concreta puede sonar mejor uno de más abajo.

## ¿Qué significan los ajustes?
**Cada modelo tiene su propia configuración** (la que publicó su autor) y se carga sola al elegirlo. En `auto` no hay nada que tocar; al terminar se imprime qué valores se usaron.

- **Calidad** — *Máxima* sube el `overlap` a 8, que es el tope útil según jarredou («normalmente no tiene sentido pasar de 8»); tarda más. *Rápida* usa el `overlap` del autor.
- **chunk_size** — el tamaño de los trozos en que se corta la canción para procesarla. Cada modelo se entrenó con uno; `auto` usa el del autor.
- **overlap** — cuánto se pisan esos trozos entre sí. Más alto = uniones más limpias y más tiempo.
- **Formato** — `auto` (por defecto) guarda cada resultado con el **mismo tipo (WAV/FLAC) y los mismos bits que su canción original, a 44 100 Hz**.
  - **Excepción, solo cuando hace falta:** si alguna pista de una canción pasa de 0 dBFS, las pistas de esa canción se guardan en **WAV FLOAT de 32 bits** y se avisa cuánto pasó. No es más calidad: en 16/24 bits esos picos se cortarían al guardar, y en FLOAT quedan intactos para que bajes la ganancia después. Puede pasar con masters muy fuertes: al bajar la canción a 44 100 Hz los picos entre muestras (*true peak*) se vuelven muestras reales, y al quitar la voz el instrumental puede quedar más alto que la mezcla. Archivos con margen (por ejemplo, exportados a −6 dB) no llegan a eso.
  - **Siempre 44 100 Hz:** casi todos los modelos (178 de 185) trabajan a 44 100 Hz y no ven nada arriba de 22 050 Hz, así que devolver una canción de 96 o 176,4 kHz a su frecuencia original solo daría un falso hi-res (archivo más grande, mismo corte en 22 kHz).
  - Si eliges un formato fijo (por ejemplo `wav PCM_24`) se respeta, y se avisa si algo se recortó.
- **TTA** — procesa la canción tres veces (normal, canales invertidos, fase invertida) y promedia. Tarda el triple y la mejora rara vez se oye.
- **extraer_instrumental** — solo en modo manual: además del stem del modelo, guarda lo que queda al restarlo de la mezcla.

Algunos modelos (SW de 6 stems, DrumSep, SCNet…) no admiten cambiar chunk/overlap: siempre usan los de su autor.

## Problemas frecuentes
- **«No hay archivos de audio»** — la carpeta de entrada debe ser una *carpeta* (no un archivo) y distingue mayúsculas: `input`, no `Input`.
- **No aparece mi Drive** — vuelve a ejecutar *1. Preparar*.
- **Se quedó sin memoria** — usa calidad *Rápida* o un `chunk_size` más pequeño.
- Al terminar: *Entorno de ejecución → Desconectar y eliminar entorno*, para no gastar tus créditos.
